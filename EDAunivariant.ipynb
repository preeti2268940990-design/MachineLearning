```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv("train.csv")
```

```python
df.sample(10)
```

#1. Categorical Data

# (a) count plot




```python
df['Embarked'].value_counts().plot(kind='bar')
```

```python
sns.countplot(x='Sex',data=df)
```

```python
sns.countplot(x='Pclass',data=df)
```

# (b) Piechart

```python
import matplotlib.pyplot as plt
```

```python
plt.pie(df['Embarked'].value_counts(),labels=df['Embarked'].value_counts().index,autopct='%1.2f%%',colors=['pink','skyblue','violet'])
```

```python
df['Survived'].value_counts().plot(kind='pie',autopct='%.2f')
```
